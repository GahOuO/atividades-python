## 9. Controle de saldo
- O programa começa com saldo de R$ 500,00. Serão realizadas 6 operações. Em cada uma, leia
o tipo (1 = depósito; 2 = saque) e o valor. Depósitos aumentam o saldo. Saques só podem
ocorrer se houver saldo suficiente; caso contrário, mostre 'Saldo insuficiente'. Ao final, mostre o
saldo.

**Antes de programar**\
O saldo é um contador, um acumulador ou o estado atual do sistema?\
Um acumulador.\
O que deve acontecer quando um saque não puder ser realizado?\
Uma mensagem de saldo insuficiente deve ser enviada.

**Resolução:**


1. O programa precisa receber o tipo de operação 6 vezes e realizá-las, caso não seja possível realizar a operação deve ser mostrada uma mensagemd de erro. No fim, o saldo deve ser mostrado.
2. Há a entrada do tipo de operação e do valor da operação 6 vezes, o armazenamento do saldo após a operação e a saída do saldo ao final.
3. Há uma estrutura de repetição finita e uma estrutura condicional.



```python
saldo = 500
for i in range(6):
  operação = int(input("Que tipo de operação você deseja realizar? (1 = depósito; 2 = saque) \n"))
  if operação == 1:
    valor = int(input("Que valor você deseja depositar? \n"))
    saldo += valor
  elif operação == 2:
    valor = int(input("Que valor você deseja sacar? \n"))
    if valor > saldo:
      print("Saldo insuficiente.")
    else:
      saldo -= valor
  else:
    print("Operação inválida")
print(f"O seu saldo final é de: R${saldo}")
```

## 12. Divisores de um número
- Leia um inteiro positivo N, mostre todos os seus divisores positivos e informe quantos divisores
ele possui.

**Resolução:**


1. O programa precisa verificar e mostrar os números que são divisores de N entre 1 e N. Se um número for divisor ele deve ser printado e o número de divisores deve ser somado em 1.
2. Há a entrada do número, o armazenamento da quantidade de divisores e a saída dos divisores e da quantidade deles.

```python
n = int(input("Digite um número inteiro positivo. \n"))
ndivisores = 0
divisor = 0
print("Divisores positivos:")
for i in range(n-1):
  divisor += 1
  if n % divisor == 0:
    print(divisor)
    ndivisores += 1
print(f"Número de divisores: {ndivisores}")
```

## 15. Desafio final - caixa de atendimento
- Um caixa inicia o dia com R$ 200,00. Leia operações até que o código 0 seja informado. Código 1
= entrada de dinheiro; código 2 = retirada. Para cada operação válida, leia o valor. Uma retirada
só é permitida se houver saldo suficiente. Ao final, informe: saldo final, quantidade de entradas,
quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total
que saiu

**Resolução:**


1. O programa precisa setar o saldo como 200. Em uma repetição sem fim definido ele deve perguntar que código o usuário deseja utilizar e realizar a devida operação. O loop deve ter fim quando o código 0 for digitado e o saldo final, quantidade de entradas, quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total que saiu devem ser armazenados.
2. Entrada com quantidade indefinida de códigos e valores, armazenamento do saldo final, quantidade de entradas, quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total que saiu, saída de mensagem caso dê erro e do saldo final, quantidade de entradas, quantidade de retiradas realizadas, quantidade de retiradas recusadas, total que entrou e total que saiu.
3. Estrutura de repetição com fim indefinido, estrutura condicional e condição de parada.




```python
saldo = 200 # representa quanto dinheiro ainda tem no caixa
entradas = 0 # representa a quantidade de vezes que entrou dinheiro
retiradas = 0 # representa a quantidade de vezes que saiu dinheiro
retiradas_recusadas = 0 # representa a quantidade de vezes que não foi possível retirar dinheiro
total_entrou = 0 # representa o dinheiro que entrou no caixa
total_saiu = 0 # representa o dinheiro que saiu no caixa
while True:
  código = int(input("Que código você deseja utilizar? (0 = Finalizar; 1 = entrada de dinheiro; 2 = retirada) \n"))
  if código == 0:
    break
  if código == 1:
    valor = int(input("Que valor você deseja depositar? \n"))
    saldo += valor
    total_entrou += valor
    entradas += 1
  elif código == 2:
    valor = int(input("Que valor você deseja retirar? \n"))
    if valor > saldo:
      print("Saldo insuficiente.")
      retiradas_recusadas += 1
    else:
      saldo -= valor
      total_saiu += valor
      retiradas += 1
  else:
    print("Operação inválida")
print(f"O seu saldo final é de: R${saldo}")
print(f"A quantidade de entradas foi: {entradas}")
print(f"A quantidade de retiradas foi: {retiradas}")
print(f"O total que entrou foi: R${total_entrou}")
print(f"O total que saiu foi: R${total_saiu}")
```
