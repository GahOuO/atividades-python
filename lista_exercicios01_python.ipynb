#5. Situação do aluno

Leia duas notas de um aluno e calcule sua média.
Informe:
1. Aprovado, se a média for maior ou igual a 7;
2. Recuperação, se a média for maior ou igual a 5 e menor que 7;
2. Reprovado, se a média for menor que 5.
- Ao final, mostre também a média calculada.

Raciocínio
1. Ler dois números.
2. Somar os dois números
3. Dividir a soma por 2 para conseguir a média.
4. Se a média for maior ou igual a 7, está aprovado.
5. Se a média for maior ou igual a 5 e menor que 7, está de recuperação.
6. Se a média for menor que 5, está reprovado.
7. Mostrar a média.

````python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
media = (nota1 + nota2) / 2
if media >= 7:
  print("Aprovado")
elif media >= 5:
  print("Recuperação")
else:
  print("Reprovado")
print("Média:", media)
````

#13. Tentativas de senha


Peça ao usuário que digite a senha.
- Enquanto a senha estiver incorreta, o programa deverá solicitar uma nova tentativa.
- Quando o usuário acertar, mostre:
```
Acesso permitido!
```

Raciocínio
1. Criar uma variável com a senha certa.
2. Usar o while, que criará um loop enquanto a situação for verdadeira.
3. Se a senha errada for inserida, será escrita uma mensagem avisando.
3. Se a senha correta for escrita, o loop terminará e será escrito "Acesso permitido!".


````python
senha_certa = "ifpi2026"
while True:
  senha = input("Digite a senha: ")
  if senha == senha_certa:
    print("Acesso permitido!")
    break
  else:
    print("Senha incorreta. Tente novamente.")
````

#14. Número secreto

Considere que o número secreto é 7.
- Peça ao usuário que tente descobrir o número.
Enquanto ele não acertar:
- se o número digitado for menor que 7, mostre "Tente um número
maior";
- se for maior que 7, mostre "Tente um número menor".
Quando acertar, mostre:
```
Parabéns! Você acertou.
```

Raciocínio
1. Definir as variáveis do número secreto e do número de tentatias.
2. Usar while para que enquanto o número secreto não for escrito, não parar o loop.
3. Usar elif e print para falar quando deve ser menor ou maior.
4. Somar 1 tentativa a mais a cada erro
5. Mostrar o número de tentativas

````python
num_secreto = 7
tentativas = 1
while True:
  numero = int(input("Tente adivinhar o número secreto: "))
  if numero == num_secreto:
    print("Parabéns! Você acertou.")
    break
  elif numero < num_secreto:
    print("Tente um número maior.")
    tentativas += 1
  else:
    print("Tente um número menor.")
    tentativas += 1
print(f"Número de tentativas: {tentativas}")
````

#15. Analisando uma sequência de números


Faça um programa que leia números inteiros repetidamente.
- O programa deverá parar quando o usuário digitar 0.
- O zero serve apenas para encerrar e não deve entrar nos cálculos.


Ao final, informe:
- quantidade de números digitados;
- quantidade de números positivos;
- quantidade de números negativos;
- soma de todos os números digitados.

Raciocínio
1. Criar todas as variáveis necessárias
2. Criar um loop que continue enquanto o número lido não for o zero.
3. Após o loop, Calcular a media, quantidade de positivos, de negativos e a soma dos numeros digitados
4. Mostrar os resultados

````python
soma = 0
quantidade = 0
menor = 0
maior = 0
pares = 0
impares = 0
while True:
  numero = int(input("Digite um número inteiro: "))
  if numero == 0:
    break
  soma += numero
  quantidade += 1
  if numero < 0:
    menor = numero
  elif numero > 0:
    maior = numero
print("Menor número:", menor)
print("Maior número:", maior)
print("Quantidade de números pares:", pares)
print("Quantidade de números ímpares:", impares)
````
